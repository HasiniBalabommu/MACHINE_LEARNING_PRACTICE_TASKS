```python
import pandas as pd
import numpy as np

# Dataset
data = {
    "Brightness":[40,50,60,10,70,60,25],
    "Saturation":[20,50,90,25,70,10,80],
    "Class":["Red","Blue","Blue","Red","Blue","Red","Blue"]
}

df = pd.DataFrame(data)

print(df)

# New sample
new_point = np.array([45,40])

print("\nNew Point:", new_point)

print("\nEuclidean Distances:\n")

for i,row in df.iterrows():
    point = np.array([row["Brightness"],row["Saturation"]])
    distance = np.sqrt(np.sum((point-new_point)**2))
    print(f"Point {point} ({row['Class']}) --> Distance = {distance:.2f}")
```
