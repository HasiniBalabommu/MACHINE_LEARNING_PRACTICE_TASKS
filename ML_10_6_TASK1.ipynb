```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

```

```python
# Sample Old Faithful Geyser dataset
data = {
    'Eruption_Duration': [3.6,1.8,3.3,2.2,4.0,2.5,4.5,1.9,3.8,2.0,
                          4.2,1.7,3.9,2.3,4.4,1.6,3.7,2.4,4.1,1.8],
    'Waiting_Time': [79,54,74,62,85,55,88,52,83,57,
                     84,51,82,60,86,53,80,58,87,56]
}
df = pd.DataFrame(data)
print(df)

X = df[['Eruption_Duration', 'Waiting_Time']]
kmeans = KMeans(n_clusters=2, random_state=42)
df['Cluster'] = kmeans.fit_predict(X)

print("\nClustered Data:\n")
print(df)
plt.figure(figsize=(8,6))

```

```python
plt.scatter(df['Eruption_Duration'],
            df['Waiting_Time'],
            c=df['Cluster'],
            cmap='viridis',
            s=100)
centers = kmeans.cluster_centers_

plt.scatter(centers[:,0],
            centers[:,1],
            color='red',
            marker='X',
            s=250,
            label='Centroids')

plt.title("K-Means Clustering of Old Faithful Geyser")
plt.xlabel("Eruption Duration (minutes)")
plt.ylabel("Waiting Time (minutes)")
plt.legend()
plt.show()
```
