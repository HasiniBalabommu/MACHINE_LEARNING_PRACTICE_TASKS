```python
import pandas as pd

# PCA Loading Matrix

loadings = pd.DataFrame({

'PC1':[0.190,0.544,0.782,0.365,0.585,0.394,0.985,0.520],
'PC2':[0.017,0.020,-0.605,0.294,0.085,-0.273,0.126,0.402],
'PC3':[0.207,0.204,0.144,0.585,0.234,0.027,-0.111,0.519]

},

index=['Climate',
       'Housing',
       'Health',
       'Crime',
       'Transportation',
       'Education',
       'Arts',
       'Recreation'])

print(loadings)

print("\nHighest Loading Variable in Each Principal Component")

for col in loadings.columns:
    max_var = loadings[col].abs().idxmax()
    max_val = loadings[col].loc[max_var]
    print(f"{col} --> {max_var} ({max_val})")
```
